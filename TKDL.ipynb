```python
brands = [f for f in os.listdir(parent_dir) if os.path.isdir(os.path.join(parent_dir, f))]
brands
```

```python
all_stats = defaultdict(float)
brand_stats = defaultdict(lambda: defaultdict(float))
```

```python
path = '20221234-20222345.Honda.1'
text = path.split('.')[1]
text
```

```python
file_pattern = r"([\d\-]+)\.([a-zA-Z]+)\.(\d+)"

for brand in os.listdir(parent_dir):
    brand_path = os.path.join(parent_dir, brand)
    if os.path.isdir(brand_path):  # Chỉ xử lý thư mục
        print(f"Processing brand directory: {brand}")  # Debug: Brand folder
        for file in os.listdir(brand_path):
            file_path = os.path.join(brand_path, file)
            if os.path.isfile(file_path):
                print(f"Processing file: {file}")  # Debug: File name
                # Tìm MSSV, tên hiệu xe, và số thứ tự ảnh từ tên file
                match = re.match(file_pattern, file)
                if match:
                    ids, car_brand, _ = match.groups()
                    ids_list = ids.split("-")
                    contribution = 1 / len(ids_list)  # Chia đều đóng góp giữa các MSSV

                    print(f"Matched IDs: {ids_list}, Brand: {car_brand}, Contribution: {contribution}")  # Debug: Match info

                    for mssv in ids_list:
                        all_stats[mssv] += contribution
                        brand_stats[mssv][car_brand] += contribution
                        print(f"Updated stats -> MSSV: {mssv}, All: {all_stats[mssv]}, Brand: {car_brand}, Contribution: {brand_stats[mssv][car_brand]}")  # Debug: Updated stats

print("\nFinal All Stats:")  # Debug: Summary of all stats
for mssv, quantity in all_stats.items():
    print(f"MSSV: {mssv}, Total Contribution: {quantity}")

print("\nFinal Brand Stats:")  # Debug: Summary of brand stats
for mssv, brands in brand_stats.items():
    for brand, quantity in brands.items():
        print(f"MSSV: {mssv}, Brand: {brand}, Contribution: {quantity}")


with open("CarDataset-1.csv", mode="w", newline="", encoding="utf-8") as file1:
    writer = csv.writer(file1)
    writer.writerow(["MSSV", "All", "Số lượng"])
    for mssv, quantity in all_stats.items():
        writer.writerow([mssv, "All", round(quantity, 2)])


with open("CarDataset-2.csv", mode="w", newline="", encoding="utf-8") as file2:
    writer = csv.writer(file2)
    writer.writerow(["MSSV", "Hiệu xe", "Số lượng"])
    for mssv, brands in brand_stats.items():
        for brand, quantity in brands.items():
            writer.writerow([mssv, brand, round(quantity, 2)])
```
